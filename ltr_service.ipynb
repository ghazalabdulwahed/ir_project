```python
import pandas as pd

# تحديد مسار الملف
file_path = r'C:\Users\ev\Documents\ir_project-main\quora_cleaned_sample.csv'

# قراءة الملف
df = pd.read_csv(file_path)

# عرض أول 5 أسطر للتأكد من نجاح العملية
print(df.head())
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

# 1. تنظيف البيانات: تحويل القيم الفارغة إلى نصوص فارغة
df['cleaned_text'] = df['cleaned_text'].fillna('')

# 2. التأكد من أن جميع القيم عبارة عن نصوص (String)
df['cleaned_text'] = df['cleaned_text'].astype(str)

# 3. الآن يمكنك تشغيل الـ TF-IDF بدون مشاكل
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(df['cleaned_text'])

print("تم تحويل النصوص بنجاح!")
```

```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

def get_top_k_tfidf(query, vectorizer, tfidf_matrix, df, k=10):
    query_vec = vectorizer.transform([query])

    scores = cosine_similarity(query_vec, tfidf_matrix).flatten()

    results = df.copy()
    results["tfidf_score"] = scores

    return results.sort_values("tfidf_score", ascending=False).head(k)
```

```python
# استدعاء الدالة وتخزين النتائج
query = "how to invest in share market"
final_results = get_top_k_tfidf(query, vectorizer, tfidf_matrix, df, k=10)

# عرض النتائج
print(final_results[['original_text', 'tfidf_score']])
```
