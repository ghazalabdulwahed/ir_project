```python
import pandas as pd
from collections import defaultdict, Counter
import time

print("Environment initialized successfully.")
```

```python
print("⏳ Loading dataset...")
start_time = time.time()

# تحميل الداتا الكبيرة
df = pd.read_pickle("processed_documents.pkl")

print(f"✔ Data loaded in {time.time() - start_time:.2f} seconds.")
print(f"Total Documents: {len(df):,}")
print("Columns available:", df.columns.tolist())

# عرض أول سطرين للتأكد من بنية البيانات
df.head(2)
```

```python
def build_day1_inverted_index(df):
    print("⏳ Processing documents and building the Inverted Index...")
    start_time = time.time()
    
    inverted_index = defaultdict(set)
    doc_lengths = {}
    doc_term_freqs = {}
    
    # استخدام zip للمرور السريع جداً على الداتا الكبيرة
    for doc_id, cleaned_text in zip(df["doc_id"], df["cleaned_text"]):
        words = str(cleaned_text).split()
        
        # 1. حفظ طول المستند (مهم لـ BM25 في اليوم 3)
        doc_lengths[doc_id] = len(words)
        
        # 2. حساب تكرار الكلمات داخل المستند (مهم لـ TF-IDF في اليوم 2)
        counts = Counter(words)
        doc_term_freqs[doc_id] = counts
        
        # 3. بناء الفهرس المعكوس
        for word in counts.keys():
            inverted_index[word].add(doc_id)
            
    print(f"✔ Done! Indexing took {time.time() - start_time:.2f} seconds.")
    return inverted_index, doc_lengths, doc_term_freqs

# تشغيل الدالة وتخزين المخرجات في الرام
inverted_index, doc_lengths, doc_term_freqs = build_day1_inverted_index(df)

print(f"Total Unique Words (Vocabulary Size): {len(inverted_index):,}")
```

```python
import math
import time

def compute_global_idf(df, inverted_index):
    print("⏳ Calculating global IDF weights for all terms...")
    start_time = time.time()
    
    N = len(df) # إجمالي عدد المستندات |D|
    idf_weights = {}
    
    for word, doc_ids in inverted_index.items():
        df_t = len(doc_ids) # عدد المستندات التي تحتوي على الكلمة
        
        # تطبيق معادلة المحاضرة بالتفصيل
        idf_weights[word] = math.log(1 + (N / df_t))
        
    print(f"✔ IDF calculation complete in {time.time() - start_time:.2f} seconds.")
    return idf_weights

# تشغيل الدالة
idf_weights = compute_global_idf(df, inverted_index)
```

```python
def compute_vsm_tfidf(doc_term_freqs, idf_weights):
    print("⏳ Generating TF-IDF sparse vectors for all documents...")
    start_time = time.time()
    
    tfidf_vectors = {}
    
    # المرور على تكرارات الكلمات الجاهزة من اليوم الأول
    for doc_id, term_counts in doc_term_freqs.items():
        tfidf_vectors[doc_id] = {}
        
        for word, count in term_counts.items():
            # 1. حساب الـ TF المخمد باللوغاريتم
            tf_smoothed = math.log(1 + count)
            
            # 2. جلب الـ IDF الخاص بالكلمة
            word_idf = idf_weights.get(word, 0)
            
            # 3. حساب الوزن النهائي حسب معادلة المحاضرة
            tfidf_vectors[doc_id][word] = tf_smoothed * word_idf
            
    print(f"✔ TF-IDF vectors generated successfully in {time.time() - start_time:.2f} seconds.")
    return tfidf_vectors

# تشغيل الدالة
tfidf_vectors = compute_vsm_tfidf(doc_term_freqs, idf_weights)
```

```python
import math
import time

def prepare_bm25_weights(df, inverted_index):
    print("⏳ Precomputing BM25 probabilistic IDF weights...")
    start_time = time.time()
    
    N = len(df) # إجمالي عدد المستندات |D|
    bm25_idf = {}
    
    for word, doc_ids in inverted_index.items():
        df_term = len(doc_ids) # عدد المستندات التي تحتوي على الكلمة (df_t)
        
        # تطبيق معادلة الـ IDF الخاصة بـ BM25 والموجودة في المراجع الأكاديمية
        numerator = N - df_term + 0.5
        denominator = df_term + 0.5
        bm25_idf[word] = math.log((numerator / denominator) + 1)
        
    print(f"✔ BM25 IDF weights prepared in {time.time() - start_time:.2f} seconds.")
    return bm25_idf

# تشغيل الدالة
bm25_idf = prepare_bm25_weights(df, inverted_index)

# حساب متوسط طول المستندات عبر كامل الـ Dataset (مهم جداً لمعادلة BM25)
avg_doc_length = sum(doc_lengths.values()) / len(df)
print(f"Global Average Document Length (avg_doc_len): {round(avg_doc_length, 2)}")
```

```python
def bm25_search_engine(
    query,
    inverted_index,
    bm25_idf,
    avg_doc_length,
    doc_lengths,
    doc_term_freqs,
    k1=1.5,   # ديناميكي: للتحكم بإشباع تكرار الكلمة
    b=0.75    # ديناميكي: للتحكم بنسبة تأثير طول المستند
):
    # 1. تنظيف وتقطيع الاستعلام (بافتراض تمرير النص بعد معالجته أو تقطيعه)
    # لتسهيل الفحص السريع، سنقوم بعمل .split() على الكلمات المجذرة مباشرة
    query_terms = str(query).lower().split()
    
    if not query_terms:
        return []
        
    # 2. Boolean Filtering: جلب المستندات المرشحة فقط التي تحتوي على الأقل على كلمة واحدة من الاستعلام
    candidate_docs = set()
    for term in query_terms:
        if term in inverted_index:
            candidate_docs.update(inverted_index[term])
            
    # 3. حساب الأوزان للمستندات المرشحة فقط
    scores = {}
    for doc_id in candidate_docs:
        doc_len = doc_lengths[doc_id]
        term_freqs = doc_term_freqs[doc_id] # تكرارات الكلمات داخل هذا المستند
        score = 0
        
        for term in query_terms:
            if term not in term_freqs:
                continue
                
            tf = term_freqs[term]
            term_idf = bm25_idf.get(term, 0)
            
            # حساب المعادلة الكسرية لـ BM25
            numerator = tf * (k1 + 1)
            denominator = tf + k1 * (1 - b + b * (doc_len / avg_doc_length))
            
            score += term_idf * (numerator / denominator)
            
        scores[doc_id] = score
        
    # 4. ترتيب النتائج تنازلياً حسب السكور
    return sorted(scores.items(), key=lambda x: x[1], reverse=True)
```

```python
# اختيار استعلام طبي تجريبي (مجذر ومناسب للداتا عندك)
test_query = "calcium channel" 

start_search = time.time()
results = bm25_search_engine(
    query=test_query,
    inverted_index=inverted_index,
    bm25_idf=bm25_idf,
    avg_doc_length=avg_doc_length,
    doc_lengths=doc_lengths,
    doc_term_freqs=doc_term_freqs,
    k1=1.5, # يمكنك تغييرها وتجربة أثرها
    b=0.75  # يمكنك تغييرها وتجربة أثرها
)
search_duration = time.time() - start_search

print(f"🔍 Search results for: '{test_query}'")
print(f"⚡ Time taken: {search_duration * 1000:.2f} milliseconds!")
print(f"Total relevant documents found: {len(results):,}\n")

print("🏆 Top 5 Matched Documents:")
for rank, (doc_id, score) in enumerate(results[:5], 1):
    print(f" Rank {rank} | Doc ID: {doc_id:<12} | BM25 Score: {score:.4f}")
```
